```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.model_selection import cross_val_score

from sklearn import datasets
```

```python
x, y = datasets.make_blobs(500, 2, centers=2, cluster_std=0.8 )
```

```python
plt.scatter(x[:, 0], x[:, 1], c=y);
```

```python
model = SVC(C=1, kernel='linear' )
model.fit(x, y)
```

```python
model.score(x, y)
```

```python
vectors = model.support_vectors_
model.support_vectors_
```

```python
len(model.support_vectors_)
```

```python
model.coef_
```

```python
w = model.coef_[0]
b = model.intercept_[0]
m = w[0]/w[1]
X = np.linspace(-1, 1) * (x[:, 0].max() - x[:, 0].min())

Y = -m * X -b / w[1]

print(w, b, m, X, Y, sep='\n')
```

```python
plt.scatter(x[:, 0], x[:, 1], c=y);
plt.scatter(vectors[:, 0], vectors[:, 1,], s= 150, facecolor='none', edgecolor='r')

plt.plot(X, Y, 'r')
# plt.axis("square")
```
