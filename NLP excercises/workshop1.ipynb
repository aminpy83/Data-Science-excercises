### 6. Define Text Preprocessing Function

This function, `text_preprocessing`, takes a text and an optional `english` flag. It performs several common NLP preprocessing steps:
- **Tokenization**: Breaks text into words.
- **Normalization**: Converts tokens to lowercase, removes punctuation and digits.
- **English Specific Steps (if `english=True`)**:
  - **Stopword Removal**: Eliminates common words (e.g., 'the', 'is').
  - **Stemming**: Reduces words to their root form (e.g., 'running' -> 'run').
  - **Lemmatization**: Reduces words to their base or dictionary form (e.g., 'better' -> 'good').
  - **POS Tagging**: Assigns grammatical categories (e.g., noun, verb) to words.
  - **Named Entity Recognition (NER)**: Identifies and classifies named entities (e.g., persons, organizations, locations).

```python
def text_preprocessing(txt, english=True):
                                                                                                    # tokenization
    tokens = word_tokenize(txt)
                                                                                                    # normalization
    lowercase = [token.lower() for token in tokens]
    no_punc = [token for token in lowercase if token not in string.punctuation]
    no_digit = [token for token in no_punc if not token.isdigit()]

    no_stop, stemmed, lemmatized, pos_tagged, ner = 5 * ['non-english text']

    if english :
        no_stop = [token for token in no_digit if token not in stopwords.words('english')]          # deleting stopwords

        stemmed = [PorterStemmer().stem(token) for token in no_stop]                                # stemming

        lemmatized = [WordNetLemmatizer().lemmatize(token) for token in no_stop]                    # lemmatization

        pos_tagged = pos_tag(no_punc)                                                               # pos tagging

        ner = ne_chunk(pos_tagged)                                                                  # name entity recognition

    procs = (tokens,
        lowercase,
        no_punc,
        no_digit,
        no_stop,
        stemmed,
        lemmatized,
        pos_tagged,
        ner)

    return procs
```

### 7. Apply Preprocessing to English Text (text1)

This cell calls the `text_preprocessing` function with `text1` (English) and prints the results of each preprocessing step, including tokenization, normalization, stopword removal, stemming, lemmatization, POS tagging, and NER.

```python
preprocessing = text_preprocessing(text1)
print(
        f'tokens:                   {preprocessing[0]}',
        f'lowercase normalization:  {preprocessing[1]}',
        f'removing punctuations:    {preprocessing[2]}',
        f'removing digits:          {preprocessing[3]}',
        f'removing stopwords:       {preprocessing[4]}',
        f'stemming:                 {preprocessing[5]}',
        f'lemmatization:            {preprocessing[6]}',
        f'part of speach tagging:   {preprocessing[7]}',
        f'name entity recognition:  {preprocessing[8]}',
        50*"--",
        sep='\n'
        )
```

### 8. Apply Preprocessing to Non-English Text (text2)

Since `english` is set to `False`, the English-specific steps (stopwords, stemming, lemmatization, POS tagging, NER) are skipped, and a placeholder `non-english text` is returned for those steps.

### 9. Regex for Text Cleaning

Regular expressions (`regex`) are powerful for searching and manipulating strings. Here, we'll use `re` to remove special characters and extra spaces from `text1`.

```python
import re

def clean_text_regex(text):
    # Remove special characters (keep letters, numbers, and spaces)
    cleaned_text = re.sub(r'[^a-zA-Z0-9\s]', '', text)
    # Replace multiple spaces with a single space
    cleaned_text = re.sub(r'\s+', ' ', cleaned_text).strip()
    return cleaned_text

fabricated_text = "Hello World! This is a test text with some @special #characters and $numbers 123. It also has   extra spaces. Let's clean it!"

cleaned_text_fabricated = clean_text_regex(fabricated_text)
print("Original fabricated text (excerpt):", fabricated_text[:200], "...")
print("Cleaned fabricated text (excerpt):", cleaned_text_fabricated[:200], "...")
```

### 10. Hazm for Persian Text Processing

Now, let's use the `hazm` library to preprocess some Persian text. We'll demonstrate tokenization, stemming, and lemmatization.

```python
from hazm import WordTokenizer, Stemmer, Lemmatizer

persian_text = "سلام دنیا! این یک متن فارسی برای پردازش با استفاده از کتابخانه هازم است. پردازش زبان طبیعی یک زمینه جذاب است."

# Tokenization
tokenizer = WordTokenizer()
persian_tokens = tokenizer.tokenize(persian_text)
print(f"Persian Tokens: {persian_tokens}")

# Stemming
stemmer = Stemmer()
persian_stemmed = [stemmer.stem(token) for token in persian_tokens]
print(f"Persian Stemmed: {persian_stemmed}")

# Lemmatization
lemmatizer = Lemmatizer()
persian_lemmatized = [lemmatizer.lemmatize(token) for token in persian_tokens]
print(f"Persian Lemmatized: {persian_lemmatized}")
```
