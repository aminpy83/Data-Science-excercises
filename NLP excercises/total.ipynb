```python
text = "Computer science is the study of computers algorithms data and software It focuses on solving problems through logical thinking programming and mathematical reasoning The field includes artificial intelligence cybersecurity data science computer networks software engineering and database systems Computer scientists design efficient systems create applications analyze information and improve the performance of technology Computer science plays an essential role in modern society by supporting communication education healthcare business transportation and scientific research As technology continues to evolve computer science remains one of the most important and rapidly growing disciplines in the world"
print(text)
```

```python
from sklearn.preprocessing import OneHotEncoder
import numpy as np

words = text.replace('.', '').replace(',', '').split()

vocabulary = sorted(list(set(words)))
word_to_index = {word: i for i, word in enumerate(vocabulary)}

# one-hot encoding
one_hot_encoded_vectors = []
for word in words:
    vector = np.zeros(len(vocabulary))
    vector[word_to_index[word]] = 1
    one_hot_encoded_vectors.append(vector)

print("Vocabulary:", vocabulary)
print("\nWord to index mapping:", word_to_index)
print("\nOne-Hot Encoded Vectors (first 5 words):\n", np.array(one_hot_encoded_vectors[:5]))
print("\nShape of One-Hot Encoded Vectors:", np.array(one_hot_encoded_vectors).shape)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# tf-idf
vectorizer_tfidf = TfidfVectorizer()
tfidf_matrix = vectorizer_tfidf.fit_transform([text])
tfidf_feature_names = vectorizer_tfidf.get_feature_names_out()

print("TF-IDF Vocabulary:", tfidf_feature_names)
print("\nTF-IDF Vector:\n", tfidf_matrix.toarray()[0])
```

```python
from sklearn.feature_extraction.text import CountVectorizer

# bag of words
vectorizer_bow = CountVectorizer()
bow_matrix = vectorizer_bow.fit_transform([text])
bow_feature_names = vectorizer_bow.get_feature_names_out()

print("Bag of Words Vocabulary:", bow_feature_names)
print("\nBag of Words Vector:\n", bow_matrix.toarray()[0])
```

```python
!pip install gensim nltk

from gensim.models import Word2Vec
from nltk.tokenize import word_tokenize, sent_tokenize
import nltk

try:
    nltk.data.find('tokenizers/punkt_tab')
except LookupError:
    nltk.download('punkt_tab')

sentences = sent_tokenize(text)
tokenized_sentences = [word_tokenize(sentence.lower()) for sentence in sentences]

# word embedding با word2vec
model = Word2Vec(tokenized_sentences, vector_size=100, window=5, min_count=1, workers=4)

print("Word2Vec model trained successfully!")
print("Vector for 'computer':", model.wv['computer'])
print("Words similar to 'science':", model.wv.most_similar('science'))

words_to_compare = ['computer', 'science']
if 'apple' in model.wv.key_to_index:
    words_to_compare.append('apple')

if len(words_to_compare) >= 3:
    try:
        print("Word that doesn't match ('computer', 'science', 'apple'):", model.wv.doesnt_match(words_to_compare))
    except KeyError as e:
        print(f"Could not find all words for doesnt_match: {e}")
else:
    print("Not enough unique words to demonstrate 'doesnt_match' with 'apple'.")
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer, PorterStemmer
from nltk.tokenize import word_tokenize
import string
import re

# اطمینان از دانلود منابع NLTK لازم
try:
    nltk.data.find('tokenizers/punkt')
except LookupError:
    nltk.download('punkt')
try:
    nltk.data.find('corpora/stopwords')
except LookupError:
    nltk.download('stopwords')
try:
    nltk.data.find('corpora/wordnet')
except LookupError:
    nltk.download('wordnet')
try:
    nltk.data.find('corpora/omw-1.4')
except LookupError:
    nltk.download('omw-1.4')

# 1. Lowercasing
processed_text = text.lower()
print("1. After Lowercasing:", processed_text[:200], "...")

# 2. Punctuation & Special Character Removal
processed_text = re.sub(f'[{re.escape(string.punctuation)}]', '', processed_text)
processed_text = re.sub(r'\d+', '', processed_text)
processed_text = re.sub(r'\s+', ' ', processed_text).strip()
print("2. After Punctuation & Special Character Removal:", processed_text[:200], "...")

# 3. Tokenization
tokenized_words = word_tokenize(processed_text)
print("3. After Tokenization (first 20 words):", tokenized_words[:20], "...")

# 4. Stop-Word Removal
stop_words = set(stopwords.words('english'))
filtered_words = [word for word in tokenized_words if word not in stop_words]
print("4. After Stop-Word Removal (first 20 words):", filtered_words[:20], "...")

# 5. Lemmatization & Stemming
lemmatizer = WordNetLemmatizer()
lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_words]
print("5.1. After Lemmatization (first 20 words):", lemmatized_words[:20], "...")

stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in filtered_words]
print("5.2. After Stemming (first 20 words):", stemmed_words[:20], "...")
```
