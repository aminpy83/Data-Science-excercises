```python
from transformers import pipeline

# sample
sample_texts = [
    "این کتاب واقعاً عالی بود، از خواندنش لذت بردم.",
    "اصلا از کیفیت این محصول راضی نیستم، خیلی بد بود.",
    "هوا امروز معمولی است، نه خیلی خوب و نه خیلی بد.",
    "پیشنهاد می‌کنم حتما این فیلم را ببینید، فوق‌العاده است.",
    "خدمات مشتریان بسیار ضعیف و ناامیدکننده بود."
]

# multilingual model
sentiment_pipeline = pipeline("sentiment-analysis", model="cardiffnlp/twitter-xlm-roberta-base-sentiment")

# Running sentiment analysis
results = sentiment_pipeline(sample_texts)

for text, result in zip(sample_texts, results):
    print(f"متن: {text}")
    print(f"برچسب: {result['label']}, امتیاز: {result['score']:.4f}")
    print("-" * 30)
```
